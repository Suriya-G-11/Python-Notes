```python
# While Loop:
# # Simple Login Example
# Write a Python program to allow a user to enter a password. 
# The user gets a maximum of 3 attempts. If the correct password is entered, print "Login successful".
# If the password is incorrect after 3 attempts, print "Sorry, your account is blocked"."

password = ""
attempts = 0

while password != 'Tech@123' and attempts <3:
    password = input("Enter your password:")
    attempts += 1
    if password!= 'Tech@123'and attempts <3:
        print('Sorry, Enter the correct password')
if password == 'Tech@123':
    print('Login Successful.')
else:
    print('Sorry, your account is blocked')
    
```
